## 1. Setup & Array Anatomy

The core object is the `ndarray`: a grid of values, **all of the same type**,
stored in a contiguous block of memory. That homogeneity + contiguity is what
makes it fast.

Key attributes every array carries:

| Attribute | Meaning |
|-----------|---------|
| `ndim`     | number of dimensions (axes) |
| `shape`    | size along each axis, as a tuple |
| `size`     | total number of elements |
| `dtype`    | data type of elements |
| `itemsize` | bytes per element |
| `nbytes`   | total bytes consumed |

```python
import numpy as np
print("NumPy version:", np.__version__)

a = np.array([[1, 2, 3],
              [4, 5, 6]])
print("array:\n", a)
print("ndim     :", a.ndim)
print("shape    :", a.shape)
print("size     :", a.size)
print("dtype    :", a.dtype)
print("itemsize :", a.itemsize, "bytes")
print("nbytes   :", a.nbytes, "bytes")
```

## 3. Data Types (dtypes)

Every array has one dtype. Choosing it well controls memory and precision.
Common ones: `int8/16/32/64`, `uint*`, `float16/32/64`, `complex64/128`, `bool_`, `str_`, `object`.

```python
x = np.array([1, 2, 3], dtype=np.int8)
print(x.dtype, x.itemsize, "byte each")

# Casting with astype (always returns a NEW array)
f = x.astype(np.float64)
print(f, f.dtype)

# Overflow is silent with fixed-width ints — be careful!
print(np.array([127], dtype=np.int8) + np.array([1], dtype=np.int8))  # wraps to -128

# Inspect / compare dtypes
print(np.issubdtype(f.dtype, np.floating))   # True
print(np.can_cast(np.int8, np.int64))        # True
```

## 4. Indexing & Slicing

Slicing syntax is `start:stop:step`. For N-D arrays, separate axes with commas.
**Slices return views** (we'll prove this in section 16).

```python
a = np.arange(10)
print(a[0], a[-1])        # first, last
print(a[2:7])             # [2 3 4 5 6]
print(a[::2])             # every other -> [0 2 4 6 8]
print(a[::-1])            # reversed

g = np.arange(1, 13).reshape(3, 4)
print("grid:\n", g)
print("row 1     :", g[1])          # whole row
print("col 2     :", g[:, 2])       # whole column
print("submatrix :\n", g[0:2, 1:3]) # rows 0-1, cols 1-2
print("elem [2,3]:", g[2, 3])       # single element
```

```python
# Ellipsis (...) fills in as many ':' as needed
cube = np.arange(24).reshape(2, 3, 4)
print(cube[..., 0].shape)   # (2, 3)  -> first slab of last axis
print(cube[0, ...].shape)   # (3, 4)

# Assignment through slices
z = np.zeros((3, 3))
z[1, :] = [7, 8, 9]
print(z)
```

## 5. Fancy & Boolean Indexing

- **Fancy indexing**: index with an array/list of integer positions.
- **Boolean indexing**: index with a mask of `True`/`False`.

Both return **copies**, not views.

```python
a = np.array([10, 20, 30, 40, 50])

# Fancy: pick positions in any order, with repeats
print(a[[0, 2, 4]])        # [10 30 50]
print(a[[4, 4, 0]])        # [50 50 10]

# 2-D fancy: pairs of (row, col)
g = np.arange(1, 13).reshape(3, 4)
rows = [0, 1, 2]
cols = [1, 2, 3]
print(g[rows, cols])        # elements (0,1),(1,2),(2,3) -> [2 7 12]
```

```python
# Boolean masking
data = np.array([3, -1, 4, -5, 9, -2])
mask = data > 0
print(mask)                 # [ True False  True False  True False]
print(data[mask])           # [3 4 9]
print(data[data < 0])       # [-1 -5 -2]

# Combine conditions with & | ~ (parentheses required!)
print(data[(data > 0) & (data < 5)])   # [3 4]

# Assign through a mask: clip negatives to 0
data[data < 0] = 0
print(data)                 # [3 0 4 0 9 0]
```

## 6. Reshaping & Dimension Manipulation

Reshape never changes data, only how it's viewed. Use `-1` to let NumPy infer one axis.

```python
a = np.arange(12)
print(a.reshape(3, 4))
print(a.reshape(2, -1))     # -1 inferred as 6
print(a.reshape(4, 3, order='F'))  # column-major fill

# Flatten vs ravel
m = np.arange(6).reshape(2, 3)
print(m.ravel())            # view when possible (fast)
print(m.flatten())          # always a copy

# Transpose
print(m.T)                  # (3, 2)
cube = np.arange(24).reshape(2, 3, 4)
print(cube.transpose(1, 0, 2).shape)  # reorder axes -> (3, 2, 4)
```

```python
# Adding / removing axes
v = np.array([1, 2, 3])
print(v[:, np.newaxis].shape)       # (3, 1) column vector
print(v[np.newaxis, :].shape)       # (1, 3) row vector
print(np.expand_dims(v, axis=0).shape)

big = np.ones((1, 3, 1))
print(np.squeeze(big).shape)        # (3,) -> drops size-1 axes
```

## 8. Broadcasting

Broadcasting lets NumPy operate on arrays of different shapes without copying data.

**The rules** (compare shapes right-to-left):
1. If dimensions differ in count, the smaller shape is padded with 1s on the left.
2. Two dimensions are compatible if they are **equal** or **one of them is 1**.
3. A size-1 dimension is stretched to match the other.

If any pair is incompatible → error.

```python
# scalar broadcasts to everything
a = np.arange(6).reshape(2, 3)
print(a + 100)

# (2,3) + (3,)  -> row vector added to each row
print(a + np.array([10, 20, 30]))

# (2,3) + (2,1) -> column vector added to each column
print(a + np.array([[100], [200]]))

# Outer product via broadcasting
col = np.arange(1, 4)[:, None]   # (3,1)
row = np.arange(1, 5)[None, :]   # (1,4)
print(col * row)                 # (3,4) multiplication table
```

## 9. Universal Functions (ufuncs)

ufuncs apply element-wise, in fast compiled C. They support broadcasting,
an optional `out=` target, and reduction methods like `.reduce` and `.accumulate`.

```python
x = np.array([1, 4, 9, 16], dtype=float)
print(np.sqrt(x))
print(np.exp([0, 1, 2]))
print(np.log(np.e ** np.arange(3)))
print(np.sin(np.linspace(0, np.pi, 5)).round(3))

# Reductions and accumulations on any binary ufunc
print(np.add.reduce(np.arange(1, 5)))      # 1+2+3+4 = 10
print(np.add.accumulate(np.arange(1, 5)))  # [1 3 6 10]
print(np.multiply.reduce(np.arange(1, 6))) # 5! = 120

# out= writes in place, avoiding a new allocation
dst = np.empty(4)
np.sqrt(x, out=dst)
print(dst)
```

## 10. Aggregations & the `axis` Argument

The single most important idea: **`axis` is the dimension that gets collapsed.**
- `axis=0` collapses rows → one result **per column**.
- `axis=1` collapses columns → one result **per row**.

```python
m = np.arange(1, 13).reshape(3, 4)
print(m)
print("sum all        :", m.sum())
print("sum axis=0 (cols):", m.sum(axis=0))   # down each column
print("sum axis=1 (rows):", m.sum(axis=1))   # across each row

print("mean :", m.mean(), "| std:", m.std().round(3))
print("min/max:", m.min(), m.max())
print("argmax (flat idx):", m.argmax())
print("cumsum:\n", m.cumsum(axis=1))

# keepdims keeps the collapsed axis as size 1 (great for broadcasting)
print("row sums keepdims:\n", m.sum(axis=1, keepdims=True))
```

## 12. Linear Algebra (`np.linalg`)

Use `@` or `np.matmul` for matrix multiplication; `*` is element-wise.

```python
A = np.array([[1, 2], [3, 4]], dtype=float)
B = np.array([[5, 6], [7, 8]], dtype=float)

print("A @ B:\n", A @ B)
print("dot  :\n", np.dot(A, B))
print("element-wise A*B:\n", A * B)

print("transpose:\n", A.T)
print("inverse:\n", np.linalg.inv(A))
print("determinant:", np.linalg.det(A))
print("trace:", np.trace(A))
print("matrix rank:", np.linalg.matrix_rank(A))
```

```python
# Solve a linear system  A x = b   (don't invert — solve!)
A = np.array([[3, 1], [1, 2]], dtype=float)
b = np.array([9, 8], dtype=float)
x = np.linalg.solve(A, b)
print("solution x:", x)                 # [2. 3.]
print("check A@x:", A @ x)              # [9. 8.]

# Eigenvalues / eigenvectors
vals, vecs = np.linalg.eig(np.array([[2, 0], [0, 3]], dtype=float))
print("eigenvalues:", vals)

# Norms and SVD
print("L2 norm:", np.linalg.norm([3, 4]))   # 5.0
U, S, Vt = np.linalg.svd(np.array([[1, 2], [3, 4], [5, 6]], dtype=float))
print("singular values:", S.round(3))
```

## 13. The Random Module

Modern NumPy prefers the **Generator** API: `rng = np.random.default_rng(seed)`.
It's faster, statistically better, and the seed gives reproducibility.

```python
rng = np.random.default_rng(42)

print("uniform [0,1):", rng.random(3).round(3))
print("integers    :", rng.integers(low=0, high=10, size=5))
print("normal      :", rng.normal(loc=0, scale=1, size=3).round(3))
print("choice      :", rng.choice(['a', 'b', 'c'], size=5))

deck = np.arange(10)
rng.shuffle(deck)                 # in place
print("shuffled    :", deck)
print("permutation :", rng.permutation(5))   # returns new array
```

## 14. Sorting, Searching & Counting

```python
a = np.array([3, 1, 4, 1, 5, 9, 2, 6])
print("sorted      :", np.sort(a))        # returns a copy
print("argsort     :", np.argsort(a))     # indices that would sort
print("a[argsort]  :", a[np.argsort(a)])  # == np.sort(a)

# Sort a 2-D array along an axis
m = np.array([[3, 1, 2], [9, 7, 8]])
print("sort axis=1 :\n", np.sort(m, axis=1))

# where: vectorized if/else
print(np.where(a > 3, a, 0))      # keep if > 3 else 0
print(np.where(a > 3))            # just the indices (as a tuple)

# nonzero, searchsorted
print("nonzero idx :", np.nonzero(a > 3))
print("insert pos  :", np.searchsorted([1, 3, 5, 7], 4))  # 2
```

```python
# Partition: fast "k smallest" without full sort
a = np.array([7, 2, 9, 4, 1, 8, 3])
print("3 smallest (unordered):", np.partition(a, 3)[:3])

# argmin / argmax, extremes ignoring NaN
print("argmin:", a.argmin(), "argmax:", a.argmax())
vals = np.array([1.0, np.nan, 3.0])
print("nanmax:", np.nanmax(vals), "nanmean:", np.nanmean(vals))
```

## 16. Copies vs. Views — the #1 Gotcha

- **Slicing** returns a **view** → writing to it changes the original.
- **Fancy / boolean indexing** returns a **copy**.
- Use `.copy()` when you need independence. Use `arr.base` / `np.shares_memory` to check.

```python
a = np.arange(10)

view = a[2:5]          # view
view[:] = 99
print("after editing view:", a)   # original changed!
print("shares memory:", np.shares_memory(a, view))

a = np.arange(10)
copy = a[[2, 3, 4]]    # fancy indexing -> copy
copy[:] = 99
print("after editing copy:", a)   # original UNCHANGED
print("shares memory:", np.shares_memory(a, copy))

# reshape may return a view; the result's .base points to the source
r = a.reshape(2, 5)
print("reshape is a view:", r.base is a)
```

## 18. File Input/Output

- `.npy` (`np.save`/`np.load`) — one array, binary, exact.
- `.npz` (`np.savez`) — multiple arrays in one file.
- Text (`np.savetxt`/`np.loadtxt`) — human-readable CSV-like.

```python
import tempfile, os
tmp = tempfile.mkdtemp()

a = np.arange(12).reshape(3, 4)

# Binary single array
np.save(os.path.join(tmp, "a.npy"), a)
print("loaded .npy:\n", np.load(os.path.join(tmp, "a.npy")))

# Multiple arrays
np.savez(os.path.join(tmp, "bundle.npz"), first=a, second=a * 10)
bundle = np.load(os.path.join(tmp, "bundle.npz"))
print("keys:", list(bundle.keys()))
print("second:\n", bundle["second"])

# Text / CSV
np.savetxt(os.path.join(tmp, "a.csv"), a, delimiter=",", fmt="%d")
print("from csv:\n", np.loadtxt(os.path.join(tmp, "a.csv"), delimiter=","))
```

## 19. Masked Arrays & Handling NaN

Real data has holes. Two tools:
- `np.nan` + the `nan*` functions (`nanmean`, `nansum`, ...) skip NaNs.
- `np.ma` masked arrays carry an explicit mask of invalid entries.

```python
data = np.array([1.0, 2.0, np.nan, 4.0, np.nan, 6.0])

print("plain mean :", data.mean())        # nan — any NaN poisons it
print("nanmean    :", np.nanmean(data))   # ignores NaN
print("nan count  :", np.isnan(data).sum())

# Fill NaNs with the column mean
filled = np.where(np.isnan(data), np.nanmean(data), data)
print("filled     :", filled.round(2))

# Masked array approach
m = np.ma.masked_invalid(data)
print("masked mean:", m.mean())
print("compressed :", m.compressed())     # only valid values
```

## 20. Vectorization & Performance

The golden rule: **push loops into NumPy.** Vectorized code is often 10–100× faster
than Python loops because the work happens in compiled C over contiguous memory.

```python
import time

n = 1_000_000
x = np.arange(n, dtype=float)

# Slow: Python loop
start = time.perf_counter()
total = 0.0
for v in x:
    total += v * v
loop_time = time.perf_counter() - start

# Fast: vectorized
start = time.perf_counter()
total_vec = (x * x).sum()
vec_time = time.perf_counter() - start

print(f"loop       : {loop_time*1000:7.1f} ms -> {total:.3e}")
print(f"vectorized : {vec_time*1000:7.1f} ms -> {total_vec:.3e}")
print(f"speedup    : {loop_time / vec_time:7.1f}x")
```

```python
# Memory layout matters: summing along the contiguous axis is faster.
big = np.ones((2000, 2000))
import time
for axis in (0, 1):
    t = time.perf_counter()
    big.sum(axis=axis)
    print(f"sum axis={axis}: {(time.perf_counter()-t)*1000:.1f} ms")

# np.einsum: one expression for many linear-algebra patterns
A = np.arange(6).reshape(2, 3)
B = np.arange(6).reshape(3, 2)
print("einsum matmul equals @:",
      np.allclose(np.einsum('ij,jk->ik', A, B), A @ B))
print("einsum trace:", np.einsum('ii->', np.eye(4) * 3))  # 12
```

**Exercise 20** — Compute the pairwise squared Euclidean distance matrix between 5 points in 2-D, with **no Python loops** (hint: broadcasting with `[:, None, :]`).

```python
# Your turn:
```

```python
# Solution
rng = np.random.default_rng(5)
pts = rng.random((5, 2))
diff = pts[:, None, :] - pts[None, :, :]    # (5,5,2)
dist2 = (diff ** 2).sum(axis=-1)            # (5,5)
print(dist2.round(3))
print("diagonal is zero:", np.allclose(np.diag(dist2), 0))
```

**Capstone C — Standardize & one-hot.** Given a feature matrix and integer labels 0..K-1: (1) z-score each column (mean 0, std 1). (2) Build the one-hot encoding of the labels with pure NumPy.

```python
# Your turn:
```

```python
# Solution
rng = np.random.default_rng(12)
X = rng.normal(5, 2, size=(6, 3))
y = np.array([0, 2, 1, 2, 0, 1])

Xz = (X - X.mean(axis=0)) / X.std(axis=0)
print("column means ~0:", Xz.mean(axis=0).round(12))
print("column stds  ~1:", Xz.std(axis=0).round(12))

K = y.max() + 1
onehot = np.eye(K, dtype=int)[y]
print("one-hot:\n", onehot)
```
